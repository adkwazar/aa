<h4> Paradoks Simpsona

```r
N = 10
x = rnorm(N)
y1 = 5*x+20 + rnorm(N)
y2 = 5*x + 0 + rnorm(N)
y3 = 5*x-20 + rnorm(N)

plot(x, y1, xlim = c(-3,10), ylim = c(-40,40), col = "red")
lines(x+3, y2, type = "p", col = "blue")
lines(x+6, y3, type = "p", col = "green")
```

```r
X = c(x, x+3, x+6)
Y = c(y1, y2, y3)
m = lm(Y~X)
print(summary(m))


plot(x, y1, xlim = c(-3,10), ylim = c(-40,40), col = "red")
lines(x+3, y2, type = "p", col = "blue")
lines(x+6, y3, type = "p", col = "green")
abline(m)
```
